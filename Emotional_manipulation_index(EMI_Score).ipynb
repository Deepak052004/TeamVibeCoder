```python
import nltk
import re
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
```

```python
EMOTIONAL_WORDS = [
    "shocking", "terrifying", "horrifying", "unbelievable",
    "outrageous", "fear", "panic", "angry", "angering",
    "disaster", "collapse", "explosive", "breaking", "flabbergasted"
]

SENSATIONAL_PHRASES = [
    "you won’t believe",
    "shocking truth",
    "what happens next",
    "this will shock you",
    "secret revealed",
    "miracle cure"
]
```

```python
def preprocess_text(text):
    text = text.lower()
    tokens = word_tokenize(text)
    return tokens
```

```python
def count_emotional_words(tokens):
    count = sum(1 for word in tokens if word in EMOTIONAL_WORDS)
    return count
```

```python
def count_sensational_phrases(text):
    text = text.lower()
    count = sum(1 for phrase in SENSATIONAL_PHRASES if phrase in text)
    return count
```

```python
def count_excess_punctuation(text):
    exclamations = text.count('!')
    question_marks = text.count('?')
    return exclamations + question_marks
```

```python
def count_all_caps_words(text):
    words = text.split()
    caps_words = [word for word in words if word.isupper() and len(word) > 2]
    return len(caps_words)
```

```python
def emotional_manipulation_index(text):
    tokens = preprocess_text(text)

    emotional_count = count_emotional_words(tokens)
    sensational_count = count_sensational_phrases(text)
    punctuation_count = count_excess_punctuation(text)
    caps_count = count_all_caps_words(text)

    # Weighted EMI formula
    emi_score = (
        (emotional_count * 0.4) +
        (sensational_count * 0.3) +
        (punctuation_count * 0.2) +
        (caps_count * 0.1)
    )

    return round(emi_score, 3)
```

```python
def interpret_emi(score):
    if score >= 2.5:
        return "High Emotional Manipulation"
    elif score >= 1.2:
        return "Moderate Emotional Manipulation"
    else:
        return "Low Emotional Manipulation"
```

```python
nltk.download('punkt_tab')

text = """
SHOCKING TRUTH!!! You won’t believe what happened next!!!
This miracle cure will cause panic across the country!!!
"""

score = emotional_manipulation_index(text)
label = interpret_emi(score)

print("EMI Score:", score)
print("Interpretation:", label)
```
