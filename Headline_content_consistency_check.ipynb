```python
# Import libraries
import nltk
import re
from nltk.corpus import stopwords
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
nltk.download('stopwords')
```

```python
def clean_text(text):
    # Convert to lowercase
    text = text.lower()

    # Remove punctuation & numbers
    text = re.sub(r'[^a-z\s]', '', text)

    # Tokenize
    words = text.split()

    # Remove stopwords
    stop_words = set(stopwords.words('english'))
    words = [word for word in words if word not in stop_words]

    return ' '.join(words)
```

```python
def headline_content_consistency(headline, content, threshold=0.5):
    # Clean the texts
    clean_headline = clean_text(headline)
    clean_content = clean_text(content)

    # TF-IDF vectorization
    vectorizer = TfidfVectorizer()
    tfidf_matrix = vectorizer.fit_transform([clean_headline, clean_content])

    # Cosine similarity
    similarity = cosine_similarity(tfidf_matrix[0:1], tfidf_matrix[1:2])[0][0]

    # Decision
    verdict = "Consistent" if similarity >= threshold else "Inconsistent"

    return similarity, verdict
```

```python
headline = "Shocking truth about vaccine revealed!"
content = """
Health officials released a routine update about vaccine safety.
No unusual side effects were reported in the recent study.
"""

score, verdict = headline_content_consistency(headline, content)

print("Consistency Score:", round(score, 3))
print("Verdict:", verdict)
```

```python
headline = "Government announces new education policy"
content = """
The government today announced a new education policy focusing on
curriculum updates and digital learning initiatives.
"""

score, verdict = headline_content_consistency(headline, content)

print("Consistency Score:", round(score, 3))
print("Verdict:", verdict)
```

```python
print("=== Headline–Content Consistency Checker ===\n")

# Take headline input
headline = input("Enter the NEWS HEADLINE:\n")

print("\n---------------------------------\n")

# Take article content input
content = input("Enter the NEWS ARTICLE CONTENT:\n")

# Run consistency check
score, verdict = headline_content_consistency(headline, content)

print("\n========== RESULT ==========")
print("Consistency Score:", round(score, 3))
print("Verdict:", verdict)
```
