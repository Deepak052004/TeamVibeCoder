```python
import nltk
import re
from nltk.tokenize import word_tokenize
```

```python
nltk.download('punkt')
```

```python
CERTAINTY_WORDS = [
    "always", "never", "guaranteed", "proven", "undeniable",
    "will", "definitely", "certainly", "must", "no doubt",
    "everyone knows", "fact", "truth"
]

UNCERTAINTY_WORDS = [
    "may", "might", "could", "possibly", "reportedly",
    "according to", "appears", "suggests", "allegedly",
    "under investigation", "likely", "unclear"
]
```

```python
def preprocess_text(text):
    text = text.lower()
    tokens = word_tokenize(text)
    return tokens
```

```python
def count_linguistic_markers(text, tokens):
    certainty_count = sum(1 for word in tokens if word in CERTAINTY_WORDS)
    uncertainty_count = sum(1 for word in tokens if word in UNCERTAINTY_WORDS)

    # Phrase-level matching
    text_lower = text.lower()
    for phrase in CERTAINTY_WORDS:
        if " " in phrase and phrase in text_lower:
            certainty_count += 1

    for phrase in UNCERTAINTY_WORDS:
        if " " in phrase and phrase in text_lower:
            uncertainty_count += 1

    return certainty_count, uncertainty_count
```

```python
def linguistic_certainty_balance(text):
    tokens = preprocess_text(text)
    certainty, uncertainty = count_linguistic_markers(text, tokens)

    total = certainty + uncertainty + 1  # +1 to avoid division by zero

    certainty_ratio = certainty / total
    uncertainty_ratio = uncertainty / total

    return round(certainty_ratio, 3), round(uncertainty_ratio, 3)
```

```python
def interpret_linguistic_balance(certainty_ratio, uncertainty_ratio):
    if certainty_ratio > 0.7:
        return "Over-Certain Language (Suspicious)"
    elif uncertainty_ratio >= 0.4:
        return "Cautious / Journalistic Language"
    else:
        return "Moderately Balanced Language"
```

```python
nltk.download('punkt_tab')

text = """
This is the TRUTH. The government WILL collapse.
Everyone knows this is a FACT and it will definitely happen.
"""

c_ratio, u_ratio = linguistic_certainty_balance(text)
label = interpret_linguistic_balance(c_ratio, u_ratio)

print("Certainty Ratio:", c_ratio)
print("Uncertainty Ratio:", u_ratio)
print("Interpretation:", label)
```

```python
text = """
According to officials, the policy may affect the economy.
Experts suggest the impact could vary depending on multiple factors.
"""

c_ratio, u_ratio = linguistic_certainty_balance(text)
label = interpret_linguistic_balance(c_ratio, u_ratio)

print("Certainty Ratio:", c_ratio)
print("Uncertainty Ratio:", u_ratio)
print("Interpretation:", label)
```
